# NBDとディリクレ分布を組み合わせたモデルからダブルジョパディを導く：導出の詳細

NBDとディリクレ分布を組み合わせたモデルのパラメータ $(M, K, \boldsymbol{\alpha})$ から、
「シェアが低いブランドは、期間浸透率と、そのブランドを買った人1人当たりの平均購買回数がともに低い」という関係が、モデルの仮定のもとでどのように導かれるかを4つのSTEPで説明する。

以下では、同じ市場内のブランドを比較するため、カテゴリーの購買を表す $M$ と $K$、およびブランドの選択確率のばらつきを表す $\alpha_0$ は共通とする。ブランドによる違いは、$\alpha_j$ と市場シェア $s_j=\alpha_j/\alpha_0$ によって表す。$M,K>0$、$0<\alpha_j<\alpha_0$ とし、$\mu_i$ と $\mathbf{p}_i$ は独立と仮定する。

---

## 目標

ブランド $j$ の**浸透率** $b_j$（期間中に1回以上購買した消費者の割合）を求める。

$$b_j = P(X_{ij} \geq 1) = 1 - P(X_{ij} = 0)$$

まず $P(X_{ij} = 0)$ を計算する。

---

## 本文の式（5.8）との対応

本文の式（5.8）は、次の順に導かれる。

| STEP | 平均する対象 | 得られる確率 |
|---|---|---|
| STEP 1 | まだ平均しない | 総購買回数 $N_i=n$ と選択確率 $p_{ij}=p$ が与えられたときの0回購買確率 $(1-p)^n$ |
| STEP 2 | $N_i$ | $P(X_{ij}=0\mid\mu_i,p_{ij}=p)=e^{-\mu_i p}$ |
| STEP 3 | $\mu_i$ | $P(X_{ij}=0\mid p_{ij}=p)=\left(K/(K+Mp)\right)^K$ |
| STEP 4 | $p_{ij}$ | $P(X_{ij}=0)={}_2F_1(K,\alpha_j;\alpha_0;-M/K)$ |

本文で最初に示す条件付き確率はSTEP 3の結果である。その後、STEP 4でこの確率を消費者間で平均し、最後に $1-P(X_{ij}=0)$ として期間浸透率 $b_j$ を求める。

---

## STEP 1　条件付き確率：$N_i = n$, $p_{ij} = p$ が既知のとき

総購買回数 $N_i = n$ が決まっていて、かつ消費者 $i$ のブランド $j$ への選択確率 $p_{ij} = p$ も決まっているとする。

このとき、$n$ 回の購買機会の中でブランド $j$ が1回も選ばれない確率は、
二項分布の「0回」の確率そのものなので：

$$P(X_{ij} = 0 \mid N_i = n,\; p_{ij} = p) = (1 - p)^n$$

---

## STEP 2　$N_i$ について平均する：ポアソン分布の確率母関数

個人の観察期間内の期待購買回数 $\mu_i$ が与えられたとき $N_i \sim \mathrm{Poisson}(\mu_i)$ なので、
$N_i$ について総和を取り、0回購買確率を平均する：

$$P(X_{ij} = 0 \mid \mu_i,\; p_{ij} = p)
= \sum_{n=0}^{\infty} (1-p)^n \cdot \frac{e^{-\mu_i} \mu_i^n}{n!}$$

$e^{-\mu_i} \sum_{n=0}^{\infty} \frac{(\mu_i(1-p))^n}{n!} = e^{-\mu_i} \cdot e^{\mu_i(1-p)}$ と整理できるから：

$$= e^{-\mu_i} \cdot e^{\mu_i(1-p)} = e^{-\mu_i p}$$

> **ポイント**：$\sum_{n=0}^\infty a^n/n! = e^a$ というテイラー展開を使っただけ。
> 結果は「$\mu_i p$ に比例して指数的に0に近づく」シンプルな式

---

## STEP 3　$\mu_i$ について平均する：ガンマ分布の積分

個人の観察期間内の期待購買回数 $\mu_i \sim \mathrm{Gamma}(K,\; K/M)$（形状 $K$、レート $K/M$）なので、
$\mu_i$ について積分し、0回購買確率を平均する。密度関数は：

$$f(\mu_i) = \frac{(K/M)^K}{\Gamma(K)} \mu_i^{K-1} e^{-(K/M)\mu_i}$$

代入して積分する：

$$P(X_{ij} = 0 \mid p_{ij} = p)
= \int_0^\infty e^{-\mu p} \cdot \frac{(K/M)^K}{\Gamma(K)} \mu^{K-1} e^{-(K/M)\mu}\; d\mu$$

$$= \frac{(K/M)^K}{\Gamma(K)} \int_0^\infty \mu^{K-1} e^{-\mu(p + K/M)}\; d\mu$$

ガンマ関数の定義 $\int_0^\infty \mu^{K-1} e^{-c\mu} d\mu = \Gamma(K)/c^K$ を使うと：

$$= \frac{(K/M)^K}{\Gamma(K)} \cdot \frac{\Gamma(K)}{(p + K/M)^K}
= \left(\frac{K/M}{p + K/M}\right)^K
= \left(\frac{K}{K + Mp}\right)^K$$

---

## STEP 4　$p_{ij}$ について平均する：ベータ分布の積分 → 超幾何関数

ディリクレ分布の周辺分布として、ブランド $j$ の選択確率 $p_{ij}$ は
ベータ分布 $\mathrm{Beta}(\alpha_j,\; \alpha_0 - \alpha_j)$ に従う。その密度関数は：

$$f(p_{ij}) = \frac{p_{ij}^{\alpha_j - 1}(1-p_{ij})^{\alpha_0 - \alpha_j - 1}}{B(\alpha_j,\; \alpha_0 - \alpha_j)},
\quad B(a,b) = \frac{\Gamma(a)\Gamma(b)}{\Gamma(a+b)}$$

$p_{ij}$ について積分し、消費者全体の0回購買確率を求める：

$$P(X_{ij} = 0)
= \int_0^1 \left(\frac{K}{K + Mp}\right)^K
  \cdot \frac{p^{\alpha_j-1}(1-p)^{\alpha_0-\alpha_j-1}}{B(\alpha_j,\;\alpha_0-\alpha_j)}\; dp$$

$\left(\dfrac{K}{K+Mp}\right)^K = \left(1 + \dfrac{M}{K}p\right)^{-K}$ と変形すると：

$$= \frac{1}{B(\alpha_j,\;\alpha_0-\alpha_j)}
  \int_0^1 \left(1 + \frac{M}{K}p\right)^{-K}
  p^{\alpha_j-1}(1-p)^{\alpha_0-\alpha_j-1}\; dp$$

ここで**ガウスの超幾何関数**の積分表現（Euler の公式）を使う：

$$\boxed{{}_2F_1(a,\, b;\, c;\, z)
= \frac{\Gamma(c)}{\Gamma(b)\Gamma(c-b)}
  \int_0^1 t^{b-1}(1-t)^{c-b-1}(1-zt)^{-a}\; dt}$$

$a = K$、$b = \alpha_j$、$c = \alpha_0$、$z = -M/K$ と対応づけると：

$$\int_0^1 \left(1+\frac{M}{K}p\right)^{-K} p^{\alpha_j-1}(1-p)^{\alpha_0-\alpha_j-1}\; dp
= \frac{\Gamma(\alpha_j)\Gamma(\alpha_0-\alpha_j)}{\Gamma(\alpha_0)}\;
  {}_2F_1\!\left(K,\,\alpha_j;\,\alpha_0;\,-\frac{M}{K}\right)$$

$B(\alpha_j, \alpha_0-\alpha_j) = \Gamma(\alpha_j)\Gamma(\alpha_0-\alpha_j)/\Gamma(\alpha_0)$ なので、分母と約分されて：

$$\boxed{P(X_{ij} = 0) = {}_2F_1\!\left(K,\, \alpha_j;\, \alpha_0;\, -\frac{M}{K}\right)}$$

---

## 最終結果：浸透率の式

$$\boxed{b_j = 1 - {}_2F_1\!\left(K,\, \alpha_j;\, \alpha_0;\, -\frac{M}{K}\right)}$$

これは、STEP 4で求めた0回購買確率を1から引いたもので、本文の式（5.8）の最終行に対応する。

---

## なぜシェアが低いブランドほど $b_j$ が低いのか

ここでも $M$、$K$、$\alpha_0$ を固定し、同じ市場内のブランドを比較する。

超幾何関数 ${}_2F_1(K, \alpha_j; \alpha_0; -M/K)$ は $\alpha_j$ が小さくなるほど**1に近づく**。
これを直感的に理解するにはSTEP 4の積分に立ち返ればよい。

$\alpha_j \to 0$ のとき、$p_{ij} \sim \mathrm{Beta}(\alpha_j, \alpha_0-\alpha_j)$ はほぼ $p=0$ に集中する。
$p=0$ を代入すると $(K/(K+M\cdot0))^K = 1$ なので：

$$P(X_{ij} = 0) \to 1 \quad \Longrightarrow \quad b_j \to 0$$

シェアとともに $b_j$ が増えることも確認する。$A=\alpha_0$、$a=\alpha_j$ とおく。$N_i=n$ のもとで $p_{ij}$ を平均すると、

$$q_n(a)=P(X_{ij}=0\mid N_i=n)=\frac{(A-a)_n}{(A)_n}$$

となる。ここで $(q)_n=q(q+1)\cdots(q+n-1)$、$(q)_0=1$ である。$0<a<A$、$n\ge1$ では、

$$q_n'(a)=-q_n(a)\sum_{k=0}^{n-1}\frac{1}{A-a+k}<0.$$

したがって、$b_n(a)=1-q_n(a)$ は $a$ とともに増える。$N_i$ の分布はブランドによらず、$P(N_i\ge1)>0$ なので、その平均 $b(a)=\mathbb{E}_N[b_N(a)]$ も増える。よって、シェア $s_j=a/A$ が低いほど $b_j$ は低い。

これは消費者全体で見た平均的な傾向である。シェアが低いブランドでも、高い選択確率 $p_{ij}$ を持つ消費者が一部に存在する可能性はある。

---

## なぜブランド $j$ を買った人1人当たりの平均購買回数 $w_j$ もシェアとともに低下するのか

ブランド $j$ を買った人1人当たりの平均購買回数は：

$$w_j = \frac{M_j}{b_j}, \quad M_j = M \cdot s_j$$

$s_j$ が小さいと、ブランドの平均購買回数 $M_j=M s_j$ と期間浸透率 $b_j$ はどちらも小さくなる。ただし、両方が小さくなるという事実だけでは、比 $w_j=M_j/b_j$ の変化は判断できない。

この関係は、$b_j$ の凹性から示せる。前節の $q_n(a)$ をもう一度微分すると、$n\ge2$ について

$$q_n''(a)=q_n(a)\left[\left(\sum_{k=0}^{n-1}\frac{1}{A-a+k}\right)^2-\sum_{k=0}^{n-1}\frac{1}{(A-a+k)^2}\right]>0.$$

したがって $b_n(a)=1-q_n(a)$ は凹関数であり、$n=1$ では線形、$n=0$ では0である。ゆえに $b(a)=\mathbb{E}_N[b_N(a)]$ も凹関数で、$b(0)=0$ を満たす。原点を通る非負の凹関数では $b(a)/a$ は非増加であるため、

$$w_j=\frac{M(a/A)}{b(a)}=\frac{M/A}{b(a)/a}$$

は $a$、したがって $s_j$ の非減少関数である。$P(N_i\ge2)>0$ なら不等号は厳密になる。このモデルでは、シェアが低いブランドほど $w_j$ も低いことがこれで示された。これは、そのブランドを買った人全員が小さな $p_{ij}$ を持つという意味ではない。

たとえば $M=4$、$K=0.8$、$\alpha_0=10$ とすると、次の値が得られる。

| 市場シェア $s_j$ | 期間浸透率 $b_j$ | 購買者1人当たりの平均購買回数 $w_j$ |
|---:|---:|---:|
| 35\% | 53.1\% | 2.64 |
| 20\% | 38.9\% | 2.06 |
| 7\% | 17.8\% | 1.57 |

$$s_j \text{ が小さい}
\;\Rightarrow\; b_j \text{ も } w_j \text{ も小さい（ダブルジョパディ）}$$

---

## まとめ

| 量 | 式 | シェアが低いと |
|---|---|---|
| ブランド平均購買回数 | $M_j = M \cdot \alpha_j/\alpha_0$ | 小さい（シェアに比例） |
| 浸透率 | $b_j = 1 - {}_2F_1(K, \alpha_j; \alpha_0; -M/K)$ | 小さい |
| 購買者1人当たりの平均購買回数 | $w_j = M_j / b_j$ | 小さい |

パラメータ $(M, K, \boldsymbol{\alpha})$ を決めると、このモデルによる3つの予測値が決まる。
