# ブランド別購買数の同時分布：導出の詳細

本文の式（5.5〜5.7）がどのように導かれるかを順に説明し、最後にブランド別購買数の周辺分布も確認する。

---

## Step 1　条件付き多項分布

総購買回数 $N_i = n_i$ とブランド選択確率ベクトル $\mathbf{p}_i = (p_{i1}, \ldots, p_{iJ})$ が与えられたとき、
ブランド別購買数ベクトル $\mathbf{X}_i = (X_{i1}, \ldots, X_{iJ})$ は多項分布に従う：

$$
P(\mathbf{X}_i = \mathbf{x}_i \mid N_i = n_i,\, \mathbf{p}_i)
= \frac{n_i!}{\prod_{j=1}^{J} x_{ij}!} \prod_{j=1}^{J} p_{ij}^{x_{ij}},
\qquad \sum_{j=1}^{J} x_{ij} = n_i
$$

---

## Step 2　$\mathbf{p}_i$ の積分消去 → ディリクレ多項分布（式 5.5）

$\mathbf{p}_i \sim \mathrm{Dirichlet}(\boldsymbol{\alpha})$ の確率密度関数は：

$$
f(\mathbf{p}_i \mid \boldsymbol{\alpha})
= \frac{\Gamma(\alpha_0)}{\prod_{j=1}^{J} \Gamma(\alpha_j)} \prod_{j=1}^{J} p_{ij}^{\alpha_j - 1},
\qquad \alpha_0 = \sum_{j=1}^{J} \alpha_j
$$

$\mathbf{p}_i$ を積分消去する：

$$
P(\mathbf{X}_i = \mathbf{x}_i \mid N_i = n_i,\, \boldsymbol{\alpha})
= \int P(\mathbf{X}_i = \mathbf{x}_i \mid n_i,\, \mathbf{p}_i)\; f(\mathbf{p}_i \mid \boldsymbol{\alpha})\; d\mathbf{p}_i
$$

多項分布の式とディリクレ密度を代入して被積分関数を整理すると：

$$
= \frac{n_i!}{\prod_j x_{ij}!} \cdot \frac{\Gamma(\alpha_0)}{\prod_j \Gamma(\alpha_j)}
\int_{\Delta^{J-1}} \prod_{j=1}^{J} p_{ij}^{x_{ij} + \alpha_j - 1}\; d\mathbf{p}_i
$$

ここで $\Delta^{J-1}$ は $J-1$ 次元単体（$\sum_j p_j = 1,\; p_j \geq 0$）。
積分部分は、パラメータ $(x_{i1}+\alpha_1, \ldots, x_{iJ}+\alpha_J)$ のディリクレ分布の正規化定数そのものだから：

$$
\int_{\Delta^{J-1}} \prod_{j=1}^{J} p_j^{(x_{ij}+\alpha_j)-1}\; d\mathbf{p}
= \frac{\prod_{j=1}^{J} \Gamma(x_{ij}+\alpha_j)}{\Gamma\!\left(\sum_j (x_{ij}+\alpha_j)\right)}
= \frac{\prod_{j=1}^{J} \Gamma(x_{ij}+\alpha_j)}{\Gamma(n_i + \alpha_0)}
$$

$\sum_j x_{ij} = n_i$、$\sum_j \alpha_j = \alpha_0$ を使った。代入すると**ディリクレ多項分布**（式 5.5）が得られる：

$$
\boxed{
P(\mathbf{X}_i = \mathbf{x}_i \mid N_i = n_i,\, \boldsymbol{\alpha})
= \frac{n_i!\; \Gamma(\alpha_0)}{\Gamma(n_i + \alpha_0)}
\prod_{j=1}^{J} \frac{\Gamma(x_{ij}+\alpha_j)}{x_{ij}!\; \Gamma(\alpha_j)}
}
$$

---

## Step 3　$N_i$ の積分消去 → ディリクレ負の多項分布（式 5.6）

$N_i \sim \mathrm{NBD}(M, K)$ の確率質量関数を $\Gamma$ 関数で表すと：

$$
P(N_i = n_i \mid M, K)
= \frac{\Gamma(n_i + K)}{\Gamma(K)\; n_i!}
\left(\frac{K}{K+M}\right)^{K}
\left(\frac{M}{K+M}\right)^{n_i}
$$

$\mathbf{x}_i$ が与えられると $n_i = \sum_j x_{ij}$ は固定されるため、周辺化は積分ではなく両式の積になる：

$$
P(\mathbf{X}_i = \mathbf{x}_i \mid M, K, \boldsymbol{\alpha})
= P(\mathbf{X}_i = \mathbf{x}_i \mid N_i = n_i,\, \boldsymbol{\alpha})
\;\times\;
P(N_i = n_i \mid M, K)
$$

式 5.5 と NBD の式を代入する：

$$
= \frac{n_i!\; \Gamma(\alpha_0)}{\Gamma(n_i + \alpha_0)}
\prod_j \frac{\Gamma(x_{ij}+\alpha_j)}{x_{ij}!\; \Gamma(\alpha_j)}
\;\times\;
\frac{\Gamma(n_i + K)}{\Gamma(K)\; n_i!}
\left(\frac{K}{K+M}\right)^{K}
\left(\frac{M}{K+M}\right)^{n_i}
$$

分子と分母の $n_i!$ が打ち消し合って消える：

$$
\boxed{
P(\mathbf{X}_i = \mathbf{x}_i \mid M, K, \boldsymbol{\alpha})
= \frac{\Gamma(n_i + K)}{\Gamma(K)}
\left(\frac{K}{K+M}\right)^{K}
\left(\frac{M}{K+M}\right)^{n_i}
\frac{\Gamma(\alpha_0)}{\Gamma(n_i + \alpha_0)}
\prod_{j=1}^{J} \frac{\Gamma(x_{ij}+\alpha_j)}{x_{ij}!\; \Gamma(\alpha_j)}
}
$$

これが**ディリクレ負の多項分布**（式 5.6）。NBD 部分がカテゴリー全体の購買頻度の異質性を捉え、ディリクレ多項部分がブランド間の選択配分を担う構造になっている。

---

## Step 4　ブランド平均購買回数 $M_j$ の導出（式 5.7）

$N_i$ と $\mathbf{p}_i$ は独立（モデルの仮定）なので：

$$
\mathbb{E}[X_{ij}]
= \mathbb{E}[N_i \cdot p_{ij}]
= \mathbb{E}[N_i] \cdot \mathbb{E}[p_{ij}]
$$

それぞれの期待値を代入する：

- $\mathbb{E}[N_i] = M$（NBD の平均の定義）
- $\mathbb{E}[p_{ij}] = \dfrac{\alpha_j}{\alpha_0}$（ディリクレ分布の平均の定義）

$$
\boxed{
M_j = \mathbb{E}[X_{ij}] = M \cdot \mathbb{E}[p_{ij}] = M \cdot \frac{\alpha_j}{\alpha_0}
}
$$

シェア $s_j = \alpha_j / \alpha_0$ が小さいブランドほど $M_j$ が小さくなる。

---

## Step 5　ブランド別購買数の周辺分布

ブランド $j$ の購買数 $X_{ij}$ について、$p=p_{ij}$ とおく。$X_{ij}\mid(N_i,p)\sim\mathrm{Binomial}(N_i,p)$ であるから、その確率母関数は

$$G_{X_j}(z)=\mathbb{E}_p\left[G_N(1-p+pz)\right]$$

である。$N_i\sim\mathrm{NBD}(M,K)$ の確率母関数

$$G_N(s)=\left[\frac{K}{K+M(1-s)}\right]^K$$

と $p\sim\mathrm{Beta}(\alpha_j,\alpha_0-\alpha_j)$ を代入し、Euler の積分表示を使うと、

$$\boxed{G_{X_j}(z)={}_2F_1\!\left(K,\alpha_j;\alpha_0;-\frac{M}{K}(1-z)\right)}.$$

特に $G_{X_j}(0)={}_2F_1(K,\alpha_j;\alpha_0;-M/K)$ であり、別ノートブックの0回購買確率と一致する。また、$G'_{X_j}(1)=M\alpha_j/\alpha_0$ であり、式（5.7）とも一致する。

この周辺分布は一般には平均 $M_j$、形状 $K$ の一変量NBDではない。実際、$s_j=\alpha_j/\alpha_0$ とすると、上式から得られる分散とそのNBDの分散との差は

$$\operatorname{Var}(X_{ij})-\left(Ms_j+\frac{(Ms_j)^2}{K}\right)=M^2\left(1+\frac1K\right)\frac{s_j(1-s_j)}{\alpha_0+1}>0$$

である（$0<s_j<1$）。したがって、同時分布がディリクレ負の多項分布になることと、各ブランドの周辺分布が同じ形状 $K$ のNBDになることは同値ではない。
