## 生存関数 $S(t)$ とハザード関数 $h(t)$ の関係

以下では、生存関数 $S(t)$ とハザード関数 $h(t)$ の基本関係を導く。寿命 $T$ は絶対連続分布に従い、考える区間で $S(t)>0$ と仮定する。

> ある時刻まで生き残る確率 $S(t)$ は、瞬間的な故障率・離脱率 $h(t)$ を時間に対して積み上げた結果で決まる。

導く基本式は次である。

$$
S(t)=\exp\left(-\int_0^t h(u)\,du\right)=\exp\bigl(-H(t)\bigr)
$$

ここで、

- $S(t)$：生存関数（時刻 $t$ まで生存する確率）
- $h(t)$：ハザード関数（瞬間的な故障率・死亡率）
- $H(t)$：累積ハザード関数

であり、

$$
H(t)=\int_0^t h(u)\,du
$$

である。

---

### 1. ハザード関数の定義

ハザード関数は、

> 時刻 $t$ まで生きていたとき、次の微小時間 $\Delta t$ の間にイベント（死亡・故障・離反など）が起こる率

である。定義式は

$$
h(t)=\lim_{\Delta t\to0}
\frac{P\bigl(t\le T<t+\Delta t\mid T\ge t\bigr)}{\Delta t}
$$

したがって、微小区間では

$$
P\bigl([t,t+\Delta t)\text{でイベント}\mid T\ge t\bigr)=h(t)\,\Delta t+o(\Delta t)
$$

と展開できる。

---

### 2. 微小時間で生き残る確率

時刻 $t$ で生存しているとき、次の微小時間 $dt$ で

- イベント発生確率：$h(t)dt$

なので、微小時間を生き残る確率は

$$
1-h(t)\Delta t+o(\Delta t)
$$

である。

---

### 3. 生存関数の微分方程式

時刻 $t+\Delta t$ まで生存するには、

1. 時刻 $t$ まで生存する
2. その後の $\Delta t$ も生存する

必要があるため、

$$
S(t+\Delta t)=S(t)\bigl(1-h(t)\Delta t+o(\Delta t)\bigr)
$$

よって

$$
\frac{S(t+\Delta t)-S(t)}{\Delta t}=-h(t)S(t)+o(1)
$$

$\Delta t\to0$ の極限を取ると、

$$
\frac{dS(t)}{dt}=-h(t)S(t)
$$

という微分方程式が得られる。

---

### 4. 微分方程式を解く

変数分離すると、

$$
\frac{1}{S(t)}\frac{dS(t)}{dt}=-h(t)
$$

すなわち

$$
\frac{dS(t)}{S(t)}=-h(t)dt
$$

積分して

$$
\int_{S(0)}^{S(t)} \frac{dS}{S}=-\int_0^t h(u)\,du
$$

より

$$
\log S(t)-\log S(0)=-\int_0^t h(u)\,du
$$

初期条件 $S(0)=1$（時刻0では必ず生存）を使うと、

$$
\log S(t)=-\int_0^t h(u)\,du
$$

指数を取って

$$
S(t)=\exp\left(-\int_0^t h(u)\,du\right)
$$

が得られる。

---

### 5. 累積ハザード $H(t)$ を使った表現

$$
H(t)=\int_0^t h(u)\,du
$$

とおけば、

$$
S(t)=e^{-H(t)}
$$

である。

直感的には、

- $h(t)$：瞬間的な危険度
- $\int_0^t h(u)\,du$：危険度の累積（危険の総量）
- $\exp\bigl(-H(t)\bigr)$：その危険をくぐり抜ける生存確率

という対応である。

例えば危険率一定 $h(t)=\lambda$ なら

$$
H(t)=\lambda t,
\qquad
S(t)=e^{-\lambda t}
$$

となり、これは指数分布の生存関数である。

> 生存確率は、瞬間ハザードを時間積分したものの「負の指数関数」で与えられる。
