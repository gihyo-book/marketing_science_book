## 補足：BG/NBD尤度関数の計算で使われる2つの寄与

### 概要

BG/NBDでは、観測履歴 $(x, t_x, T)$ に対して次の2つの可能性を足し合わせる。ここで $x$ は初回購買後の追加購買回数、$t_x$ は最後の追加購買時点、$T$ は観測期間長である。

| 寄与 | 内容 |
|------|------|
| 生存項 | 観測終了時点 $T$ でも顧客がアクティブである |
| 離反項 | 最後の購買 $t_x$ の後から観測終了時点 $T$ までの間に離反した |

`lifetimes.BetaGeoFitter` は Fader et al. (2005) の閉形式対数尤度を使う。パラメータに依存しない因子を除いた尤度は次のように整理できる。

$$
L(r,\alpha,a,b\mid x,t_x,T)
=
\frac{\Gamma(r+x)}{\Gamma(r)}
\alpha^r
\frac{B(a,b+x)}{B(a,b)}
\left[
\frac{1}{(\alpha+T)^{r+x}}
+
\mathbf{1}_{x>0}
\frac{a}{b+x-1}
\frac{1}{(\alpha+t_x)^{r+x}}
\right].
$$

### 各項の意味

第1項

$$
\frac{1}{(\alpha+T)^{r+x}}
$$

は、観測終了時点でも生存している場合の寄与である。

第2項

$$
\mathbf{1}_{x>0}
\frac{a}{b+x-1}
\frac{1}{(\alpha+t_x)^{r+x}}
$$

は、最後の購買後に離反した場合の寄与である。$x=0$ の顧客では最後の追加購買後に離反したという情報がないため、指示関数 $\mathbf{1}_{x>0}$ によってこの項は消える。なお、分数 $a/(b+x-1)$ は $x>0$ のときだけ評価する。

### 生存確率との対応

上の2項のうち「生存項」が全体に占める割合が、生存確率となる。

生存確率は「生存項 ÷ 全体（生存項＋離反項）」として求められる。このとき、尤度の係数

$$
\frac{\Gamma(r+x)}{\Gamma(r)}
\alpha^r
\frac{B(a,b+x)}{B(a,b)}
$$

は分子・分母の両方に共通して現れるため、約分されて消える。残るのは括弧内の比だけである。

$$
P(\text{alive}\mid x,t_x,T)
=
\frac{
\dfrac{1}{(\alpha+T)^{r+x}}
}{
\dfrac{1}{(\alpha+T)^{r+x}}
+
\mathbf{1}_{x>0}
\dfrac{a}{b+x-1}
\dfrac{1}{(\alpha+t_x)^{r+x}}
}
=
\left[
1+
\mathbf{1}_{x>0}
\frac{a}{b+x-1}
\left(\frac{\alpha+T}{\alpha+t_x}\right)^{r+x}
\right]^{-1}.
$$

この式から、`frequency=0` の顧客では生存確率が 1.0 になることも確認できる。これは「まだ追加購買がないため、最後の追加購買後に急に購買が途絶えた」という離反シグナルを観測していない、というBG/NBD上の扱いである。
