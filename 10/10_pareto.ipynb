# Pareto/NBDの尤度関数にガウスの超幾何関数が現れる理由

## 1. 全体像

Pareto/NBDモデルでは、顧客ごとに2つの潜在変数を仮定する。

* $\lambda$：購買レート
* $\mu$：離反レート

どちらも直接観測できないため、尤度計算ではこれらを積分消去する。

$$
L=
\int_0^\infty\int_0^\infty
L(\lambda,\mu)f(\lambda)f(\mu)
\,d\lambda\,d\mu
$$

通常、購買レートと離反レートには次のガンマ分布を仮定する。

$$
\lambda \sim \mathrm{Gamma}(r,\alpha),\quad
\mu \sim \mathrm{Gamma}(s,\beta)
$$

rateパラメータ表示では、たとえば

$$
f(\lambda)=
\frac{\alpha^r}{\Gamma(r)}
\lambda^{r-1}e^{-\alpha\lambda}
$$

である。

---

## 2. 観測データと2つの尤度項

ある顧客について、次の観測値を考える。

* $x$：初回購買後の追加購買回数
* $t_x$：最後の追加購買時刻（$x=0$ では $t_x=0$ とする）
* $T$：観測終了時刻

Pareto/NBDでは、顧客の状態を次の2通りに分けて尤度を作る。

1. 観測終了時点 $T$ でもまだ生存している
2. 最後の購買時刻 $t_x$ 以降、$T$ までのどこかで離反した

この2つの項を足し合わせることが、1人の顧客の尤度の基本構造である。

---

## 3. 「まだ生存中」の項

顧客が $T$ まで生存している場合、購買過程の部分は比例定数を除くと

$$
\lambda^x e^{-\lambda T}
$$

である。ここに $\lambda$ のガンマ分布を掛けると、

$$
\lambda^{r+x-1}e^{-(\alpha+T)\lambda}
$$

となる。したがって、$\lambda$ に関する積分はガンマ積分により

$$
\int_0^\infty
\lambda^{r+x-1}e^{-(\alpha+T)\lambda}
\,d\lambda
=
\frac{\Gamma(r+x)}{(\alpha+T)^{r+x}}
$$

と閉形式で求まる。

同様に、$T$ まで生存する確率 $e^{-\mu T}$ と $\mu$ のガンマ分布を掛けると、

$$
\int_0^\infty
\mu^{s-1}e^{-(\beta+T)\mu}
\,d\mu
=
\frac{\Gamma(s)}{(\beta+T)^s}
$$

となる。

したがって、「まだ生存中」の項は比例定数を除いて

$$
\frac{\Gamma(r+x)}{(\alpha+T)^{r+x}}
\cdot
\frac{\Gamma(s)}{(\beta+T)^s}
$$

と表せる。ここでは $\lambda$ と $\mu$ の積分が分離できるため、通常のガンマ積分だけで済む。

---

## 4. 「途中で離反した」の項

次に、顧客が最後の購買時刻 $t_x$ 以降、観測終了時点 $T$ までのどこかで離反した場合を考える。離反時刻を $\tau$ とすると、

$$
t_x \le \tau \le T
$$

である。

時刻 $\tau$ で離反する密度は

$$
\mu e^{-\mu\tau}
$$

であり、時刻 $\tau$ までに $x$ 回購買する部分は比例定数を除いて

$$
\lambda^x e^{-\lambda\tau}
$$

と書ける。

このとき $\lambda$ と $\mu$ をガンマ分布で積分消去すると、次の形が残る。

$$
\int_{t_x}^{T}
\frac{1}{(\alpha+\tau)^{r+x}}
\frac{1}{(\beta+\tau)^{s+1}}
\,d\tau
$$

これがPareto/NBDでガウスの超幾何関数が現れる核心である。

---

## 5. なぜこの積分が難しいのか

「まだ生存中」の項では、$\lambda$ の積分と $\mu$ の積分を分離できた。しかし、「途中で離反した」の項では、同じ積分変数 $\tau$ が

$$
\alpha+\tau,
\quad
\beta+\tau
$$

の両方に入る。そのため、

$$
\int f(\tau)g(\tau)\,d\tau
\neq
\int f(\tau)\,d\tau
\int g(\tau)\,d\tau
$$

であり、単純なガンマ積分には分解できない。

つまり問題は、

$$
\int
(A+\tau)^{-m}(B+\tau)^{-n}
\,d\tau
$$

という「2つの一次式のべき乗の積」の積分である。この形が、ガウスの超幾何関数につながる。

---

## 6. ガウスの超幾何関数

ガウスの超幾何関数は、次の級数で定義される。

$$
{}_2F_1(a,b;c;z)
=
\sum_{n=0}^{\infty}
\frac{(a)_n(b)_n}{(c)_n}
\frac{z^n}{n!}
$$

ここで、$(a)_n$ はポッホハマー記号である。

$$
(a)_n=a(a+1)\cdots(a+n-1),\quad (a)_0=1
$$

${}_2F_1(a,b;c;z)$ という記法は、

* $a,b$：上側パラメータ
* $c$：下側パラメータ
* $z$：変数

を表す。級数定義で $(a)_n,(b)_n$ は分子側、$(c)_n$ は分母側に現れるため、$a,b$ と $c$ をセミコロンで分けて書く。

---

## 7. オイラー積分表示

Pareto/NBDとの対応で重要なのは、ガウスの超幾何関数のオイラー積分表示である。

$$
{}_2F_1(a,b;c;z)
=
\frac{\Gamma(c)}{\Gamma(b)\Gamma(c-b)}
\int_0^1
 t^{b-1}(1-t)^{c-b-1}(1-zt)^{-a}
\,dt
$$

この積分表示には $\operatorname{Re}(c)>\operatorname{Re}(b)>0$ などの条件が必要であり、それ以外の領域では解析接続によって定義する。

この式は、ベータ関数

$$
B(x,y)=\int_0^1 t^{x-1}(1-t)^{y-1}\,dt
=\frac{\Gamma(x)\Gamma(y)}{\Gamma(x+y)}
$$

と、一般化二項定理

$$
(1-x)^{-a}=
\sum_{n=0}^{\infty}
\frac{(a)_n}{n!}x^n
$$

から導ける。

要点は、超幾何関数が「一次式のべき乗を含む積分」を表すための関数だということである。

---

## 8. Pareto/NBDの積分との対応

Pareto/NBDで残る積分を

$$
I=
\int_{t_x}^{T}
(\alpha+\tau)^{-m}(\beta+\tau)^{-n}
\,d\tau
$$

と書く。ただし、

$$
m=r+x,\quad n=s+1
$$

である。

$q=m+n-1>0$、$\delta=\beta-\alpha$ とおき、変数変換

$$
y=\frac{1}{\alpha+\tau}
$$

を使う。$d\tau=-y^{-2}dy$ であり、$\beta+\tau=(1+\delta y)/y$ なので、

$$
I=
\int_{1/(\alpha+T)}^{1/(\alpha+t_x)}
y^{q-1}(1+\delta y)^{-n}
\,dy
$$

となる。この被積分関数の原始関数は

$$
F(y)=\frac{y^q}{q}{}_2F_1(n,q;q+1;-\delta y)
$$

である。したがって、

$$\boxed{I=F\left(\frac{1}{\alpha+t_x}\right)-F\left(\frac{1}{\alpha+T}\right)}$$

となり、ガウスの超幾何関数が現れることが明示的に示された。$\alpha=\beta$ なら $\delta=0$ であり、${}_2F_1(\cdot;0)=1$ となって通常のべき積分に帰着する。

---

## 9. なぜ数値的に扱いにくいのか

ガウスの超幾何関数は級数

$$
{}_2F_1(a,b;c;z)
=
\sum_{n=0}^{\infty}
\frac{(a)_n(b)_n}{(c)_n}
\frac{z^n}{n!}
$$

で定義され、この級数は $|z|<1$ で絶対収束する。$|z|=1$ での収束はパラメータに依存し、級数の収束円外の値は解析接続で定義する。$z$ が1に近いと $z^n$ がなかなか小さくならず、多くの項が必要になる。

たとえば、

$$
0.999^{100}\approx 0.905
$$

であり、100乗してもほとんど小さくならない。

そのため、

* 計算時間が長くなる
* 丸め誤差が蓄積する
* 大きな項どうしの差で桁落ちが起こる

といった問題が生じやすい。

---

## 10. BG/NBDとの違い

Pareto/NBDでは、離反時刻 $\tau$ が連続時間で発生すると仮定する。そのため、

$$
\tau \in [t_x,T]
$$

について積分する必要があり、その結果としてガウスの超幾何関数が現れる。

一方、BG/NBDでは、離反は「購買直後にベルヌーイ試行で発生する」と仮定する。離反を購買イベント単位の離散的な過程として扱うため、連続時間の離反時刻 $\tau$ に関する積分が不要になる。

その結果、BG/NBDでは超幾何関数を含まない、より扱いやすい尤度関数が得られる。

---

## 11. まとめ

Pareto/NBDでガウスの超幾何関数が現れる理由は、次のように整理できる。

3. 購買レート $\lambda$ と離反レート $\mu$ は潜在変数なので積分消去する。
4. 「まだ生存中」の項では、$\lambda$ と $\mu$ の積分が分離でき、通常のガンマ積分で閉形式が得られる。
5. 「途中で離反した」の項では、離反時刻 $\tau$ を $t_x$ から $T$ まで積分する必要がある。
6. その結果、次の形の積分が現れる。

$$
\int_{t_x}^{T}
\frac{1}{(\alpha+\tau)^{r+x}}
\frac{1}{(\beta+\tau)^{s+1}}
\,d\tau
$$

7. この積分は単純なガンマ積分には分解できず、整理するとガウスの超幾何関数 ${}_2F_1(a,b;c;z)$ を含む形になる。

この数値計算の扱いにくさが、実務でBG/NBDがよく使われる理由の一つである。
