# 補足： BG/NBD：潜在パラメータの周辺化（式変形の詳細）

本ノートブックでは、BG/NBDモデルの尤度導出における2つの積分（購買レート $\lambda_i$ の周辺化と離反確率 $\theta_i$ の周辺化）を説明する。以下では $r,\alpha,a,b>0$、$x\in\{0,1,2,\ldots\}$、$z\ge0$ とする。


## 1. 購買レート $\lambda_i$ の周辺化

### 設定

- 購買レートの個人間混合分布：$\lambda_i \sim \text{Gamma}(r, \alpha)$（形状パラメータ $r$、レートパラメータ $\alpha$）
- PDFは次の通りである。

$$
f(\lambda) = \frac{\alpha^r}{\Gamma(r)} \lambda^{r-1} e^{-\alpha\lambda}, \quad \lambda > 0
$$

- 尤度の $\lambda$ に依存する部分：$\lambda^x e^{-\lambda z}$
  - 生存項では $z = T$、離反項では $z = t_x$

求めたい積分は次式である。

$$
I_\lambda = \int_0^\infty \lambda^x e^{-\lambda z} \cdot \frac{\alpha^r}{\Gamma(r)} \lambda^{r-1} e^{-\alpha\lambda} \, d\lambda
$$


### Step 1：定数を外に出す

$\frac{\alpha^r}{\Gamma(r)}$ は $\lambda$ に依存しない定数なので積分の外に出せる。

$$
I_\lambda = \frac{\alpha^r}{\Gamma(r)} \int_0^\infty \lambda^x e^{-\lambda z} \cdot \lambda^{r-1} e^{-\alpha\lambda} \, d\lambda
$$


### Step 2：$\lambda$ のべき乗をまとめる

$$
\lambda^x \cdot \lambda^{r-1} = \lambda^{x + r - 1} = \lambda^{(r+x)-1}
$$


### Step 3：指数部分をまとめる

$$
e^{-\lambda z} \cdot e^{-\alpha\lambda} = e^{-(\alpha + z)\lambda}
$$


### Step 4：被積分関数を整理する

Step 2・3 をまとめると、

$$
I_\lambda = \frac{\alpha^r}{\Gamma(r)} \int_0^\infty \lambda^{(r+x)-1} e^{-(\alpha+z)\lambda} \, d\lambda
$$


### Step 5：ガンマ積分の公式を適用する

**ガンマ積分の公式：**

$$
\int_0^\infty \lambda^{n-1} e^{-c\lambda} \, d\lambda = \frac{\Gamma(n)}{c^n}, \quad n > 0,\ c > 0
$$

ここで $n = r + x$、$c = \alpha + z$ として適用する。

$$
\int_0^\infty \lambda^{(r+x)-1} e^{-(\alpha+z)\lambda} \, d\lambda = \frac{\Gamma(r+x)}{(\alpha+z)^{r+x}}
$$


### Step 6：結果をまとめる

$$
I_\lambda
= \frac{\alpha^r}{\Gamma(r)} \cdot \frac{\Gamma(r+x)}{(\alpha+z)^{r+x}}
= \frac{\Gamma(r+x)}{\Gamma(r)} \cdot \frac{\alpha^r}{(\alpha+z)^{r+x}}
$$

これが本文の式である。

$$
\boxed{\int_0^\infty \lambda^x e^{-\lambda z} \cdot \frac{\alpha^r}{\Gamma(r)}\lambda^{r-1}e^{-\alpha\lambda}\,d\lambda = \frac{\Gamma(r+x)}{\Gamma(r)}\cdot\frac{\alpha^r}{(\alpha+z)^{r+x}}}
$$

**直感：** ポアソン過程の尤度核 $e^{-\lambda z} \lambda^x$ とガンマ混合分布を掛けると、ガンマ―ポアソン共役性と同じ代数により、レートが $\alpha+z$ のガンマ積分になる。$z$ が大きいほど分母が大きくなり、寄与は小さくなる。ここでガンマ分布はベイズ推論上の事前分布ではなく、顧客間異質性を表す混合分布である。

## 2. 離反確率 $\theta_i$ の周辺化

### 設定

- 離反確率の個人間混合分布：$\theta_i \sim \text{Beta}(a, b)$
- PDFは次の通りである。

$$
f(\theta) = \frac{\theta^{a-1}(1-\theta)^{b-1}}{B(a,b)}, \quad 0 < \theta < 1
$$

- ベータ関数の定義：$B(a, b) = \int_0^1 \theta^{a-1}(1-\theta)^{b-1}\,d\theta = \dfrac{\Gamma(a)\Gamma(b)}{\Gamma(a+b)}$

尤度における $\theta$ 依存部分は2種類ある。離反項は $x\ge1$ の場合に限って現れる。

| ケース | $\theta$ 依存部分 |
|--------|------------------|
| 生存項（$x$ 回購買後も生存） | $(1-\theta)^x$ |
| 離反項（$x$ 回目の購買後に離反） | $\theta(1-\theta)^{x-1}$ |


### 積分 A：生存項の $\theta$ 周辺化

$$
I_A = \int_0^1 (1-\theta)^x \cdot \frac{\theta^{a-1}(1-\theta)^{b-1}}{B(a,b)} \, d\theta
$$

**Step A-1：** 定数 $\frac{1}{B(a,b)}$ を外に出す。

$$
I_A = \frac{1}{B(a,b)} \int_0^1 (1-\theta)^x \cdot \theta^{a-1}(1-\theta)^{b-1} \, d\theta
$$

**Step A-2：** $(1-\theta)^x \cdot (1-\theta)^{b-1} = (1-\theta)^{b+x-1}$ とまとめる。

$$
I_A = \frac{1}{B(a,b)} \int_0^1 \theta^{a-1}(1-\theta)^{(b+x)-1} \, d\theta
$$

**Step A-3：** ベータ関数の定義 $\int_0^1 \theta^{a-1}(1-\theta)^{c-1}d\theta = B(a,c)$ を、$c = b+x$ として適用する。

$$
I_A = \frac{1}{B(a,b)} \cdot B(a, b+x) = \frac{B(a,b+x)}{B(a,b)}
$$

$$
\boxed{\int_0^1 (1-\theta)^x \cdot \frac{\theta^{a-1}(1-\theta)^{b-1}}{B(a,b)}\,d\theta = \frac{B(a,b+x)}{B(a,b)}}
$$


### 積分 B：離反項の $\theta$ 周辺化

$$
I_B = \int_0^1 \theta(1-\theta)^{x-1} \cdot \frac{\theta^{a-1}(1-\theta)^{b-1}}{B(a,b)} \, d\theta
$$

**Step B-1：** 定数 $\frac{1}{B(a,b)}$ を外に出す。

$$
I_B = \frac{1}{B(a,b)} \int_0^1 \theta \cdot \theta^{a-1} \cdot (1-\theta)^{x-1}(1-\theta)^{b-1} \, d\theta
$$

**Step B-2：** べき乗をまとめる。

$$
\theta \cdot \theta^{a-1} = \theta^a = \theta^{(a+1)-1}
$$
$$
(1-\theta)^{x-1}(1-\theta)^{b-1} = (1-\theta)^{b+x-2} = (1-\theta)^{(b+x-1)-1}
$$

$$
I_B = \frac{1}{B(a,b)} \int_0^1 \theta^{(a+1)-1}(1-\theta)^{(b+x-1)-1} \, d\theta
$$

**Step B-3：** ベータ関数の定義を $B(a+1, b+x-1)$ として適用する。

$$
I_B = \frac{B(a+1, b+x-1)}{B(a,b)}
$$

**Step B-4：** $B(a+1, b+x-1)$ を $B(a, b+x)$ で表す。

ベータ関数の性質 $B(a+1, c) = \dfrac{a}{a+c} B(a, c)$ を使う（$c = b+x-1$ とする）。

$$
B(a+1, b+x-1) = \frac{a}{a + (b+x-1)} B(a, b+x-1)
$$

さらに $B(a, b+x-1)$ と $B(a, b+x)$ の関係として、$B(a, c) = \dfrac{c-1}{a+c-1} B(a, c-1)$ を使う（$c = b+x$ とする）。

$$
B(a, b+x-1) = \frac{a+b+x-1}{b+x-1} B(a, b+x)
$$

よって、

$$
B(a+1,b+x-1)
= \frac{a}{a+b+x-1} \cdot \frac{a+b+x-1}{b+x-1} B(a,b+x)
= \frac{a}{b+x-1} B(a,b+x)
$$

**Step B-5：** 代入して整理する。

$$
I_B = \frac{1}{B(a,b)} \cdot \frac{a}{b+x-1} B(a,b+x) = \frac{a}{b+x-1} \cdot \frac{B(a,b+x)}{B(a,b)}
$$

$$
\boxed{\int_0^1 \theta(1-\theta)^{x-1} \cdot \frac{\theta^{a-1}(1-\theta)^{b-1}}{B(a,b)}\,d\theta = \frac{a}{b+x-1}\cdot\frac{B(a,b+x)}{B(a,b)}}
$$

## 3. 尤度関数への代入

$\lambda_i$ と $\theta_i$ を同時に周辺化した結果を代入する。生存項と離反項それぞれについて $I_\lambda$（$z=T$ または $z=t_x$）と $I_A$・$I_B$ を掛け合わせ、共通因子 $\dfrac{\Gamma(r+x)}{\Gamma(r)}\alpha^r \dfrac{B(a,b+x)}{B(a,b)}$ をくくり出すと、

$$
L(r,\alpha,a,b \mid x,t_x,T)
=
\frac{\Gamma(r+x)}{\Gamma(r)}
\alpha^r
\frac{B(a,b+x)}{B(a,b)}
\left[
    \frac{1}{(\alpha+T)^{r+x}}
    + \mathbf{1}_{x>0}
    \frac{a}{b+x-1}
    \frac{1}{(\alpha+t_x)^{r+x}}
\right]
$$

となる（Fader et al. 2005, Eq. 6）。同論文の Eq. 7 は顧客ごとの対数尤度を標本全体で合計した式である。離反項の分数は $x>0$ のときだけ評価する。

> **注意：** ここで使う $\lambda^x e^{-\lambda z}$ は、購買時点を含むポアソン過程の尤度核であり、回数だけのポアソン確率質量関数ではない。回数分布として書く場合に現れる $z^x/x!$ はデータのみに依存するため、パラメータ $(r,\alpha,a,b)$ の尤度から除いても最尤推定値は変わらない。
