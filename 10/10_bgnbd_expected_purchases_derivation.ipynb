# BG/NBDモデルにおける将来購買回数の期待値の導出

このノートでは、Fader et al. (2005) の付録に示されている次の期待値を導出する。

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,r,\alpha,a,b\}
$$

これは、観測期間 $[0,T]$ において初回購買後の追加購買回数が $X=x$、最後の追加購買時刻が $t_x$ だった顧客について、観測終了後の将来期間 $(T,T+t]$ に期待される購買回数を表す。

---

## 1. 記号と前提

BG/NBDモデルでは、顧客ごとに次の潜在パラメータを仮定する。

$$
\lambda: \text{購買レート}
$$

$$
p: \text{各購買直後に離反する確率}
$$

本文では離反確率を $\theta_i$ と書いているが、Fader et al. (2005) では $p$ と書かれている。このノートでは論文に合わせて $p$ を使う。

購買レートと離反確率の母集団分布は、

$$
\lambda \sim \mathrm{Gamma}(r,\alpha)
$$

$$
p \sim \mathrm{Beta}(a,b)
$$

である。ここでは $r,\alpha,a,b>0$、$t\ge0$ とし、ガンマ分布をrateパラメータ表示で書く。

$$
f(\lambda \mid r,\alpha)
=
\frac{\alpha^r}{\Gamma(r)}
\lambda^{r-1}e^{-\alpha\lambda}
$$

ベータ分布の密度は、

$$
f(p \mid a,b)
=
\frac{p^{a-1}(1-p)^{b-1}}{B(a,b)}
$$

である。

観測データは、

$$
(X=x,t_x,T)
$$

で表す。$x\in\{0,1,2,\ldots\}$ は初回購買後の追加購買回数、$t_x$ は最後の追加購買時刻、$T$ は観測期間の長さである。これは Fader et al. (2005) と `lifetimes` の `frequency` に共通する定義である。

また、

$$
\delta_{x>0}
=
\begin{cases}
1 & (x>0)\\
0 & (x=0)
\end{cases}
$$

を指示関数とする。

---

## 2. 個人レベルの尤度

まず、$\lambda$ と $p$ が既知であると仮定する。

観測履歴 $(X=x,t_x,T)$ は、次の2つの可能性から生じる。

1. 観測終了時点 $T$ でも顧客がアクティブである
2. $x>0$ のとき、最後の購買時刻 $t_x$ の直後に離反した

1つ目の場合、$x$ 回の購買が起こり、各購買後に離反せず、さらに $T$ まで購買が発生しない。比例定数を除いた寄与は、

$$
(1-p)^x \lambda^x e^{-\lambda T}
$$

である。

2つ目の場合、最初の $x-1$ 回の購買後には離反せず、$x$ 回目の購買直後に離反する。比例定数を除いた寄与は、

$$
p(1-p)^{x-1}\lambda^x e^{-\lambda t_x}
$$

である。ただし、この項は $x>0$ のときだけ存在する。

したがって、個人レベルの尤度は、

$$
L(\lambda,p \mid X=x,t_x,T)
=
(1-p)^x\lambda^x e^{-\lambda T}
+
\delta_{x>0}p(1-p)^{x-1}\lambda^x e^{-\lambda t_x}
$$

である。

---

## 3. 観測終了時点でアクティブである確率

$\lambda$ と $p$ が既知のとき、観測終了時点 $T$ で顧客がアクティブである確率を求める。

観測履歴のうち「$T$ でもアクティブである」場合の寄与は、

$$
(1-p)^x\lambda^x e^{-\lambda T}
$$

であった。

観測履歴全体の尤度は、

$$
L(\lambda,p \mid X=x,t_x,T)
$$

である。

したがって、条件付き確率の定義より、

$$
P(\text{active at }T \mid X=x,t_x,T,\lambda,p)
=
\frac{(1-p)^x\lambda^x e^{-\lambda T}}
{L(\lambda,p \mid X=x,t_x,T)}
$$

となる。

これは論文付録の式(A2)に対応する。

---

## 4. アクティブである場合の将来購買回数

次に、顧客が時点 $T$ でアクティブであるとわかっている場合、将来 $t$ 期間の期待購買回数を求める。

BG/NBDでは、アクティブな顧客は購買レート $\lambda$ のポアソン過程で購買する。そして各購買直後に確率 $p$ で離反する。

将来時点 $s$ でまだアクティブである確率を考える。購買が起こるたびに確率 $p$ で離反するので、実質的には「離反を引き起こす購買イベント」がレート $\lambda p$ のポアソン過程として発生する。

したがって、時点 $s$ まで離反していない確率は、

$$
P(\text{active at }T+s \mid \text{active at }T,\lambda,p)
=
e^{-\lambda ps}
$$

である。

時点 $T+s$ における瞬間的な期待購買率は、

$$
\lambda \cdot e^{-\lambda ps}
$$

である。よって、将来 $t$ 期間の期待購買回数は、

$$
\mathbb{E}\{Y(t) \mid \text{active at }T,\lambda,p\}
=
\int_0^t \lambda e^{-\lambda ps}\,ds
$$

である。

この積分を計算する。

$$
\int_0^t \lambda e^{-\lambda ps}\,ds
=
\lambda \int_0^t e^{-\lambda ps}\,ds
$$

$$
=
\lambda
\left[
\frac{-1}{\lambda p}e^{-\lambda ps}
\right]_{0}^{t}
$$

$$
=
\lambda
\left(
\frac{-1}{\lambda p}e^{-\lambda pt}
-
\frac{-1}{\lambda p}e^{0}
\right)
$$

$$
=
\lambda
\left(
\frac{1}{\lambda p}
-
\frac{1}{\lambda p}e^{-\lambda pt}
\right)
$$

$$
=
\frac{1}{p}\left(1-e^{-\lambda pt}\right)
$$

したがって、

$$
\mathbb{E}\{Y(t) \mid \text{active at }T,\lambda,p\}
=
\frac{1}{p}\left(1-e^{-\lambda pt}\right)
$$

である。これは論文付録の式(A1)に対応する。

---

## 5. 個人レベルの条件付き期待値

観測履歴 $(X=x,t_x,T)$ が与えられたとき、$\lambda,p$ を既知として将来購買回数の期待値を求める。

将来購買が発生するためには、まず観測終了時点 $T$ で顧客がアクティブでなければならない。

したがって、

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,\lambda,p\}
=
P(\text{active at }T \mid X=x,t_x,T,\lambda,p)
\cdot
\mathbb{E}\{Y(t) \mid \text{active at }T,\lambda,p\}
$$

である。

先ほど求めた2つの式を代入する。

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,\lambda,p\}
=
\frac{(1-p)^x\lambda^x e^{-\lambda T}}
{L(\lambda,p \mid X=x,t_x,T)}
\cdot
\frac{1}{p}\left(1-e^{-\lambda pt}\right)
$$

分子を展開すると、

$$
=
\frac{p^{-1}(1-p)^x\lambda^x e^{-\lambda T}
-
p^{-1}(1-p)^x\lambda^x e^{-\lambda T}e^{-\lambda pt}}
{L(\lambda,p \mid X=x,t_x,T)}
$$

指数部分をまとめると、

$$
e^{-\lambda T}e^{-\lambda pt}
=
e^{-\lambda(T+pt)}
$$

なので、

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,\lambda,p\}
=
\frac{
 p^{-1}(1-p)^x\lambda^x e^{-\lambda T}
-
 p^{-1}(1-p)^x\lambda^x e^{-\lambda(T+pt)}
}
{L(\lambda,p \mid X=x,t_x,T)}
$$

となる。これは論文付録の式(A3)に対応する。

---

## 6. 母集団パラメータだけが既知の場合

実際には、個人ごとの $\lambda$ と $p$ は観測されない。観測されるのは購買履歴 $(X=x,t_x,T)$ であり、推定されるのは母集団パラメータ $(r,\alpha,a,b)$ である。

そのため、次の期待値を計算する。

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,r,\alpha,a,b\}
$$

これは、$\lambda,p$ の事後分布で個人レベルの期待値を平均したものである。

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,r,\alpha,a,b\}
=
\int_0^1\int_0^\infty
\mathbb{E}\{Y(t) \mid X=x,t_x,T,\lambda,p\}
 f(\lambda,p \mid r,\alpha,a,b,X=x,t_x,T)
\,d\lambda\,dp
$$

これが論文付録の式(A4)である。

---

## 7. ベイズの定理による事後分布

ベイズの定理より、$\lambda,p$ の事後分布は、

$$
f(\lambda,p \mid r,\alpha,a,b,X=x,t_x,T)
=
\frac{
L(\lambda,p \mid X=x,t_x,T)
f(\lambda\mid r,\alpha)
f(p\mid a,b)
}
{L(r,\alpha,a,b \mid X=x,t_x,T)}
$$

である。

ここで分母は、$\lambda,p$ を周辺化した顧客レベルの尤度である。

$$
L(r,\alpha,a,b \mid X=x,t_x,T)
=
\int_0^1\int_0^\infty
L(\lambda,p \mid X=x,t_x,T)
f(\lambda\mid r,\alpha)
f(p\mid a,b)
\,d\lambda\,dp
$$

これは論文付録の式(A5)である。

---

## 8. 事後平均への代入

式(A4)に式(A3)と式(A5)を代入する。

まず、式(A4)は、

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,r,\alpha,a,b\}
=
\int_0^1\int_0^\infty
\mathbb{E}\{Y(t) \mid X=x,t_x,T,\lambda,p\}
 f(\lambda,p \mid r,\alpha,a,b,X=x,t_x,T)
\,d\lambda\,dp
$$

であった。

式(A3)より、

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,\lambda,p\}
=
\frac{
 p^{-1}(1-p)^x\lambda^x e^{-\lambda T}
-
 p^{-1}(1-p)^x\lambda^x e^{-\lambda(T+pt)}
}
{L(\lambda,p \mid X=x,t_x,T)}
$$

である。

式(A5)より、

$$
f(\lambda,p \mid r,\alpha,a,b,X=x,t_x,T)
=
\frac{
L(\lambda,p \mid X=x,t_x,T)
f(\lambda\mid r,\alpha)
f(p\mid a,b)
}
{L(r,\alpha,a,b \mid X=x,t_x,T)}
$$

である。

両者を掛けると、分子・分母の $L(\lambda,p \mid X=x,t_x,T)$ が打ち消し合う。

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,\lambda,p\}
 f(\lambda,p \mid r,\alpha,a,b,X=x,t_x,T)
$$

$$
=
\frac{
\left[
 p^{-1}(1-p)^x\lambda^x e^{-\lambda T}
-
 p^{-1}(1-p)^x\lambda^x e^{-\lambda(T+pt)}
\right]
 f(\lambda\mid r,\alpha)f(p\mid a,b)
}
{L(r,\alpha,a,b \mid X=x,t_x,T)}
$$

したがって、

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,r,\alpha,a,b\}
=
\frac{A-B}{L(r,\alpha,a,b \mid X=x,t_x,T)}
$$

となる。ただし、

$$
A=
\int_0^1\int_0^\infty
p^{-1}(1-p)^x\lambda^x e^{-\lambda T}
 f(\lambda\mid r,\alpha)f(p\mid a,b)
\,d\lambda\,dp
$$

$$
B=
\int_0^1\int_0^\infty
p^{-1}(1-p)^x\lambda^x e^{-\lambda(T+pt)}
 f(\lambda\mid r,\alpha)f(p\mid a,b)
\,d\lambda\,dp
$$

である。これが論文付録の式(A6)の構造である。

---

## 9. 分母の周辺尤度

後で使うために、周辺尤度を確認しておく。

個人レベル尤度は、

$$
L(\lambda,p \mid X=x,t_x,T)
=
(1-p)^x\lambda^x e^{-\lambda T}
+
\delta_{x>0}p(1-p)^{x-1}\lambda^x e^{-\lambda t_x}
$$

である。

これを $\lambda,p$ で周辺化すると、

$$
L(r,\alpha,a,b \mid X=x,t_x,T)
$$

$$
=
\int_0^1\int_0^\infty
\left[
(1-p)^x\lambda^x e^{-\lambda T}
+
\delta_{x>0}p(1-p)^{x-1}\lambda^x e^{-\lambda t_x}
\right]
 f(\lambda\mid r,\alpha)f(p\mid a,b)
\,d\lambda\,dp
$$

である。

第1項について、$\lambda$ 側は、

$$
\int_0^\infty
\lambda^x e^{-\lambda T}
\frac{\alpha^r}{\Gamma(r)}\lambda^{r-1}e^{-\alpha\lambda}
\,d\lambda
$$

$$
=
\frac{\alpha^r}{\Gamma(r)}
\int_0^\infty
\lambda^{r+x-1}e^{-(\alpha+T)\lambda}
\,d\lambda
$$

ガンマ積分

$$
\int_0^\infty u^{c-1}e^{-du}\,du=\frac{\Gamma(c)}{d^c}
$$

を用いると、

$$
=
\frac{\alpha^r}{\Gamma(r)}
\frac{\Gamma(r+x)}{(\alpha+T)^{r+x}}
$$

である。

$p$ 側は、

$$
\int_0^1
(1-p)^x
\frac{p^{a-1}(1-p)^{b-1}}{B(a,b)}
\,dp
$$

$$
=
\frac{1}{B(a,b)}
\int_0^1
p^{a-1}(1-p)^{b+x-1}
\,dp
$$

$$
=
\frac{B(a,b+x)}{B(a,b)}
$$

である。

したがって第1項は、

$$
\frac{\Gamma(r+x)}{\Gamma(r)}
\frac{\alpha^r}{(\alpha+T)^{r+x}}
\frac{B(a,b+x)}{B(a,b)}
$$

である。

第2項について、$\lambda$ 側は、

$$
\int_0^\infty
\lambda^x e^{-\lambda t_x}
\frac{\alpha^r}{\Gamma(r)}\lambda^{r-1}e^{-\alpha\lambda}
\,d\lambda
$$

$$
=
\frac{\Gamma(r+x)}{\Gamma(r)}
\frac{\alpha^r}{(\alpha+t_x)^{r+x}}
$$

である。

$p$ 側は、

$$
\int_0^1
p(1-p)^{x-1}
\frac{p^{a-1}(1-p)^{b-1}}{B(a,b)}
\,dp
$$

$$
=
\frac{1}{B(a,b)}
\int_0^1
p^a(1-p)^{b+x-2}
\,dp
$$

$$
=
\frac{B(a+1,b+x-1)}{B(a,b)}
$$

である。

ここで、

$$
B(a+1,b+x-1)
=
\frac{a}{b+x-1}B(a,b+x)
$$

を示す。

左辺は、

$$
B(a+1,b+x-1)
=
\frac{\Gamma(a+1)\Gamma(b+x-1)}{\Gamma(a+b+x)}
$$

である。ガンマ関数の性質より、

$$
\Gamma(a+1)=a\Gamma(a)
$$

また、

$$
\Gamma(b+x)=(b+x-1)\Gamma(b+x-1)
$$

なので、

$$
\Gamma(b+x-1)=\frac{\Gamma(b+x)}{b+x-1}
$$

である。したがって、

$$
B(a+1,b+x-1)
=
\frac{a\Gamma(a)}{1}
\frac{\Gamma(b+x)}{b+x-1}
\frac{1}{\Gamma(a+b+x)}
$$

$$
=
\frac{a}{b+x-1}
\frac{\Gamma(a)\Gamma(b+x)}{\Gamma(a+b+x)}
$$

$$
=
\frac{a}{b+x-1}B(a,b+x)
$$

である。

したがって第2項は、

$$
\delta_{x>0}
\frac{\Gamma(r+x)}{\Gamma(r)}
\frac{\alpha^r}{(\alpha+t_x)^{r+x}}
\frac{a}{b+x-1}
\frac{B(a,b+x)}{B(a,b)}
$$

である。

以上より、

$$
L(r,\alpha,a,b \mid X=x,t_x,T)
=
\frac{\Gamma(r+x)}{\Gamma(r)}
\alpha^r
\frac{B(a,b+x)}{B(a,b)}
\left[
\frac{1}{(\alpha+T)^{r+x}}
+
\delta_{x>0}
\frac{a}{b+x-1}
\frac{1}{(\alpha+t_x)^{r+x}}
\right]
$$

である。これは論文の式(6)と同じ式である。

## 10. Aの計算（まず $a>1$ とする）

次に、

$$
A=
\int_0^1\int_0^\infty
p^{-1}(1-p)^x\lambda^x e^{-\lambda T}
 f(\lambda\mid r,\alpha)f(p\mid a,b)
\,d\lambda\,dp
$$

を計算する。以下のように $A$ と $B$ を別々の積分として扱う計算は、$p=0$ 近傍で積分可能となる $a>1$ を仮定する。$0<a\le1$ の場合は、第17節で差 $A-B$ を直接扱う。

まず密度を代入する。

$$
A=
\int_0^1\int_0^\infty
p^{-1}(1-p)^x\lambda^x e^{-\lambda T}
\frac{\alpha^r}{\Gamma(r)}\lambda^{r-1}e^{-\alpha\lambda}
\frac{p^{a-1}(1-p)^{b-1}}{B(a,b)}
\,d\lambda\,dp
$$

$p$ に関する項、$\lambda$ に関する項、定数を整理する。

$$
A=
\frac{\alpha^r}{\Gamma(r)B(a,b)}
\int_0^1
p^{a-2}(1-p)^{b+x-1}
\,dp
\int_0^\infty
\lambda^{r+x-1}e^{-(\alpha+T)\lambda}
\,d\lambda
$$

$\lambda$ 積分は、

$$
\int_0^\infty
\lambda^{r+x-1}e^{-(\alpha+T)\lambda}
\,d\lambda
=
\frac{\Gamma(r+x)}{(\alpha+T)^{r+x}}
$$

である。

$p$ 積分は、

$$
\int_0^1
p^{a-2}(1-p)^{b+x-1}
\,dp
$$

である。ベータ関数の定義

$$
B(u,v)=\int_0^1 t^{u-1}(1-t)^{v-1}\,dt
$$

と比べると、

$$
u-1=a-2 \quad \Rightarrow \quad u=a-1
$$

$$
v-1=b+x-1 \quad \Rightarrow \quad v=b+x
$$

なので、

$$
\int_0^1
p^{a-2}(1-p)^{b+x-1}
\,dp
=
B(a-1,b+x)
$$

である。

したがって、

$$
A=
\frac{B(a-1,b+x)}{B(a,b)}
\frac{\Gamma(r+x)\alpha^r}{\Gamma(r)(\alpha+T)^{r+x}}
$$

である。これは論文付録の式(A7)である。

この式は $B(a-1,b+x)$ を通常のベータ積分として用いるため $a>1$ を必要とする。ただし、これは $A$ 単独の有限性に対する条件であり、有限期間 $t$ の将来購買回数の期待値に必要な条件ではない。

---

## 11. Bの計算：ガンマ積分まで

次に、

$$
B=
\int_0^1\int_0^\infty
p^{-1}(1-p)^x\lambda^x e^{-\lambda(T+pt)}
 f(\lambda\mid r,\alpha)f(p\mid a,b)
\,d\lambda\,dp
$$

を計算する。

密度を代入する。

$$
B=
\int_0^1\int_0^\infty
p^{-1}(1-p)^x\lambda^x e^{-\lambda(T+pt)}
\frac{\alpha^r}{\Gamma(r)}\lambda^{r-1}e^{-\alpha\lambda}
\frac{p^{a-1}(1-p)^{b-1}}{B(a,b)}
\,d\lambda\,dp
$$

項をまとめる。

$$
B=
\frac{\alpha^r}{\Gamma(r)B(a,b)}
\int_0^1
p^{a-2}(1-p)^{b+x-1}
\left[
\int_0^\infty
\lambda^{r+x-1}e^{-\lambda(\alpha+T+pt)}
\,d\lambda
\right]
\,dp
$$

内側の $\lambda$ 積分は、ガンマ積分より、

$$
\int_0^\infty
\lambda^{r+x-1}e^{-\lambda(\alpha+T+pt)}
\,d\lambda
=
\frac{\Gamma(r+x)}{(\alpha+T+pt)^{r+x}}
$$

である。

したがって、

$$
B=
\frac{\Gamma(r+x)\alpha^r}{\Gamma(r)B(a,b)}
\int_0^1
p^{a-2}(1-p)^{b+x-1}(\alpha+T+pt)^{-(r+x)}
\,dp
$$

となる。

---

## 12. Bの計算：変数変換

残った積分は、

$$
\int_0^1
p^{a-2}(1-p)^{b+x-1}(\alpha+T+pt)^{-(r+x)}
\,dp
$$

である。

ここで、

$$
q=1-p
$$

とおく。すると、

$$
p=1-q
$$

$$
dp=-dq
$$

である。

積分範囲は、

$$
p=0 \Rightarrow q=1
$$

$$
p=1 \Rightarrow q=0
$$

となる。

したがって、

$$
\int_0^1
p^{a-2}(1-p)^{b+x-1}(\alpha+T+pt)^{-(r+x)}
\,dp
$$

$$
=
\int_1^0
(1-q)^{a-2}q^{b+x-1}\{\alpha+T+(1-q)t\}^{-(r+x)}(-dq)
$$

積分範囲を $0$ から $1$ に直すと、

$$
=
\int_0^1
q^{b+x-1}(1-q)^{a-2}\{\alpha+T+t-tq\}^{-(r+x)}
\,dq
$$

ここで、

$$
\alpha+T+t-tq
=
(\alpha+T+t)\left(1-\frac{t}{\alpha+T+t}q\right)
$$

である。したがって、

$$
\{\alpha+T+t-tq\}^{-(r+x)}
=
(\alpha+T+t)^{-(r+x)}
\left(1-\frac{t}{\alpha+T+t}q\right)^{-(r+x)}
$$

となる。

これを代入すると、

$$
\int_0^1
p^{a-2}(1-p)^{b+x-1}(\alpha+T+pt)^{-(r+x)}
\,dp
$$

$$
=
(\alpha+T+t)^{-(r+x)}
\int_0^1
q^{b+x-1}(1-q)^{a-2}
\left(1-\frac{t}{\alpha+T+t}q\right)^{-(r+x)}
\,dq
$$

である。

したがって、

$$
B=
\frac{\Gamma(r+x)\alpha^r}{\Gamma(r)B(a,b)(\alpha+T+t)^{r+x}}
\int_0^1
q^{b+x-1}(1-q)^{a-2}
\left(1-\frac{t}{\alpha+T+t}q\right)^{-(r+x)}
\,dq
$$

となる。

---

## 13. オイラー積分表示との対応

ガウスの超幾何関数のオイラー積分表示は、

$$
{}_2F_1(A,B;C;z)
=
\frac{\Gamma(C)}{\Gamma(B)\Gamma(C-B)}
\int_0^1
u^{B-1}(1-u)^{C-B-1}(1-zu)^{-A}
\,du
$$

である。

これを変形すると、

$$
\int_0^1
u^{B-1}(1-u)^{C-B-1}(1-zu)^{-A}
\,du
=
\frac{\Gamma(B)\Gamma(C-B)}{\Gamma(C)}
{}_2F_1(A,B;C;z)
$$

である。

ベータ関数を使えば、

$$
\frac{\Gamma(B)\Gamma(C-B)}{\Gamma(C)}=B(B,C-B)
$$

なので、

$$
\int_0^1
u^{B-1}(1-u)^{C-B-1}(1-zu)^{-A}
\,du
=
B(B,C-B){}_2F_1(A,B;C;z)
$$

である。

いま求めたい積分は、

$$
\int_0^1
q^{b+x-1}(1-q)^{a-2}
\left(1-\frac{t}{\alpha+T+t}q\right)^{-(r+x)}
\,dq
$$

である。

オイラー積分表示と対応させるために、

$$
u=q
$$

$$
A=r+x
$$

$$
B=b+x
$$

$$
C-B-1=a-2
$$

とおく。すると、

$$
C-B=a-1
$$

であり、

$$
C=a-1+B=a-1+b+x=a+b+x-1
$$

である。また、

$$
z=\frac{t}{\alpha+T+t}
$$

である。

したがって、

$$
\int_0^1
q^{b+x-1}(1-q)^{a-2}
\left(1-\frac{t}{\alpha+T+t}q\right)^{-(r+x)}
\,dq
$$

$$
=
B(b+x,a-1)
{}_2F_1\left(r+x,b+x;a+b+x-1;\frac{t}{\alpha+T+t}\right)
$$

ベータ関数は対称なので、

$$
B(b+x,a-1)=B(a-1,b+x)
$$

である。

したがって、

$$
B=
\frac{B(a-1,b+x)}{B(a,b)}
\frac{\Gamma(r+x)\alpha^r}{\Gamma(r)(\alpha+T+t)^{r+x}}
{}_2F_1\left(r+x,b+x;a+b+x-1;\frac{t}{\alpha+T+t}\right)
$$

である。これは論文付録の式(A8)である。

---

## 14. A-Bを整理する

式(A7)より、

$$
A=
\frac{B(a-1,b+x)}{B(a,b)}
\frac{\Gamma(r+x)\alpha^r}{\Gamma(r)(\alpha+T)^{r+x}}
$$

である。

式(A8)より、

$$
B=
\frac{B(a-1,b+x)}{B(a,b)}
\frac{\Gamma(r+x)\alpha^r}{\Gamma(r)(\alpha+T+t)^{r+x}}
{}_2F_1\left(r+x,b+x;a+b+x-1;\frac{t}{\alpha+T+t}\right)
$$

である。

したがって、

$$
A-B
=
\frac{B(a-1,b+x)}{B(a,b)}
\frac{\Gamma(r+x)\alpha^r}{\Gamma(r)}
\left[
\frac{1}{(\alpha+T)^{r+x}}
-
\frac{1}{(\alpha+T+t)^{r+x}}
{}_2F_1\left(r+x,b+x;a+b+x-1;\frac{t}{\alpha+T+t}\right)
\right]
$$

である。

---

## 15. 分母で割る

求めたい期待値は、

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,r,\alpha,a,b\}
=
\frac{A-B}{L(r,\alpha,a,b \mid X=x,t_x,T)}
$$

である。

分母は、

$$
L(r,\alpha,a,b \mid X=x,t_x,T)
=
\frac{\Gamma(r+x)}{\Gamma(r)}
\alpha^r
\frac{B(a,b+x)}{B(a,b)}
\left[
\frac{1}{(\alpha+T)^{r+x}}
+
\delta_{x>0}
\frac{a}{b+x-1}
\frac{1}{(\alpha+t_x)^{r+x}}
\right]
$$

であった。

したがって、分子と分母の共通因子

$$
\frac{\Gamma(r+x)}{\Gamma(r)}\alpha^r\frac{1}{B(a,b)}
$$

は打ち消し合う。

残るのは、

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,r,\alpha,a,b\}
$$

$$
=
\frac{
B(a-1,b+x)
\left[
\frac{1}{(\alpha+T)^{r+x}}
-
\frac{1}{(\alpha+T+t)^{r+x}}
{}_2F_1\left(r+x,b+x;a+b+x-1;\frac{t}{\alpha+T+t}\right)
\right]
}
{
B(a,b+x)
\left[
\frac{1}{(\alpha+T)^{r+x}}
+
\delta_{x>0}
\frac{a}{b+x-1}
\frac{1}{(\alpha+t_x)^{r+x}}
\right]
}
$$

である。

---

## 16. ベータ関数比の整理

次に、

$$
\frac{B(a-1,b+x)}{B(a,b+x)}
$$

を整理する。

ベータ関数の定義より、

$$
B(a-1,b+x)
=
\frac{\Gamma(a-1)\Gamma(b+x)}{\Gamma(a+b+x-1)}
$$

$$
B(a,b+x)
=
\frac{\Gamma(a)\Gamma(b+x)}{\Gamma(a+b+x)}
$$

である。したがって、

$$
\frac{B(a-1,b+x)}{B(a,b+x)}
=
\frac{\Gamma(a-1)\Gamma(b+x)}{\Gamma(a+b+x-1)}
\cdot
\frac{\Gamma(a+b+x)}{\Gamma(a)\Gamma(b+x)}
$$

$$
=
\frac{\Gamma(a-1)}{\Gamma(a)}
\cdot
\frac{\Gamma(a+b+x)}{\Gamma(a+b+x-1)}
$$

ガンマ関数の性質より、

$$
\Gamma(a)=(a-1)\Gamma(a-1)
$$

なので、

$$
\frac{\Gamma(a-1)}{\Gamma(a)}=\frac{1}{a-1}
$$

である。また、

$$
\Gamma(a+b+x)=(a+b+x-1)\Gamma(a+b+x-1)
$$

なので、

$$
\frac{\Gamma(a+b+x)}{\Gamma(a+b+x-1)}=a+b+x-1
$$

である。

したがって、

$$
\frac{B(a-1,b+x)}{B(a,b+x)}
=
\frac{a+b+x-1}{a-1}
$$

である。

---

## 17. 最終式への変形

前節の結果を使うと、期待値は、

$$
\mathbb{E}\{Y(t) \mid X=x,t_x,T,r,\alpha,a,b\}
$$

$$
=
\frac{a+b+x-1}{a-1}
\cdot
\frac{
\frac{1}{(\alpha+T)^{r+x}}
-
\frac{1}{(\alpha+T+t)^{r+x}}
{}_2F_1\left(r+x,b+x;a+b+x-1;\frac{t}{\alpha+T+t}\right)
}
{
\frac{1}{(\alpha+T)^{r+x}}
+
\delta_{x>0}
\frac{a}{b+x-1}
\frac{1}{(\alpha+t_x)^{r+x}}
}
$$

である。

分子・分母に $(\alpha+T)^{r+x}$ を掛ける。

まず分子は、

$$
(\alpha+T)^{r+x}
\left[
\frac{1}{(\alpha+T)^{r+x}}
-
\frac{1}{(\alpha+T+t)^{r+x}}
{}_2F_1\left(r+x,b+x;a+b+x-1;\frac{t}{\alpha+T+t}\right)
\right]
$$

$$
=
1
-
\left(\frac{\alpha+T}{\alpha+T+t}\right)^{r+x}
{}_2F_1\left(r+x,b+x;a+b+x-1;\frac{t}{\alpha+T+t}\right)
$$

である。

分母は、

$$
(\alpha+T)^{r+x}
\left[
\frac{1}{(\alpha+T)^{r+x}}
+
\delta_{x>0}
\frac{a}{b+x-1}
\frac{1}{(\alpha+t_x)^{r+x}}
\right]
$$

$$
=
1
+
\delta_{x>0}
\frac{a}{b+x-1}
\left(\frac{\alpha+T}{\alpha+t_x}\right)^{r+x}
$$

である。

したがって、最終的に、

$$
\boxed{
\mathbb{E}\{Y(t) \mid X=x,t_x,T,r,\alpha,a,b\}
=
\frac{\frac{a+b+x-1}{a-1}
\left[
1-
\left(\frac{\alpha+T}{\alpha+T+t}\right)^{r+x}
{}_2F_1\left(r+x,b+x;a+b+x-1;\frac{t}{\alpha+T+t}\right)
\right]}
{1+
\delta_{x>0}
\frac{a}{b+x-1}
\left(\frac{\alpha+T}{\alpha+t_x}\right)^{r+x}}
}
$$

が得られる。

これがFader et al. (2005) の付録に示されている、BG/NBDモデルにおける将来購買回数の条件付き期待値である。ここまでの $A$ と $B$ を分ける導出は $a>1$ で直接正当化される。

### $0<a\le1$ の場合

$0<a\le1$ では $A$ と $B$ はそれぞれ発散するため、別々の期待値として解釈してはならない。一方、その差は

$$A-B=\frac{\Gamma(r+x)\alpha^r}{\Gamma(r)B(a,b)}\int_0^1 p^{a-2}(1-p)^{b+x-1}\left[(\alpha+T)^{-(r+x)}-(\alpha+T+pt)^{-(r+x)}\right]dp$$

と一つの積分にまとめられる。角括弧内は $p\to0$ で $O(p)$ であるため、被積分関数は $O(p^{a-1})$ となり、すべての $a>0$ で積分可能である。したがって、上の閉形式は解析接続により $a>0$、$a\ne1$ に拡張できる。$a=1$ の見かけ上の特異点は除去可能であり、差の積分または $a\to1$ の極限を数値的に安定な方法で評価する必要がある。

## 18. この式が計算しているもの

この式は、次の量を計算している。

$$
\text{過去に }(X=x,t_x,T)\text{ という購買履歴を持つ顧客が、将来 }t\text{ 期間に何回購買するかの期待値}
$$

単なる「購買レート $\times$ 期間」ではない。

この式は、

3. その顧客の購買レート $\lambda$ はどれくらい高そうか
4. その顧客の離反確率 $p$ はどれくらい高そうか
5. 観測終了時点 $T$ でまだアクティブである可能性はどれくらいか
6. 将来期間中にいつ離反する可能性があるか

をすべて統合して、将来購買回数の期待値を計算している。

`lifetimes` の `conditional_expected_number_of_purchases_up_to_time()` が計算している量も、この条件付き期待値である。
