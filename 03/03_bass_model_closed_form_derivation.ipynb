# バスモデルの解析解の導出

第3章の式 (3.9) から式 (3.10) を得るまでを、変数分離と部分分数分解に着目して数式で追う。


## 1. 出発点

累積採用者数 $N(t)$ について、バスモデルの常微分方程式を

$$

\frac{dN(t)}{dt} = \left(p + q \frac{N(t)}{M}\right)(M - N(t)),\qquad N(0)=0

$$

とおく。ここで $p$ は外部影響係数、$q$ は内部影響係数、$M$ は潜在市場規模である。

この方程式を解いて、

$$

N(t)=M \cdot \frac{1-e^{-(p+q)t}}{1+\frac{q}{p}e^{-(p+q)t}}

$$

を導く。


## 2. 市場比率への正規化

$x(t)=N(t)/M$ とおくと $N(t)=Mx(t)$ なので、

$$

\frac{dN(t)}{dt}=M\frac{dx(t)}{dt}

$$

である。これを元の方程式に代入すると

$$

M\frac{dx(t)}{dt} = \left(p + qx(t)\right)M(1-x(t))

$$

となるから、両辺を $M$ で割って

$$

\frac{dx(t)}{dt} = (p+qx(t))(1-x(t))

$$

を得る。以下では $x=x(t)$ と略記する。


## 3. 変数分離と部分分数分解

前節の式 $\dfrac{dx}{dt} = (p+qx)(1-x)$ を変数分離すると

$$
\frac{dx}{(p+qx)(1-x)} = dt
$$

となる。この左辺を積分するために、被積分関数を部分分数分解する。未定係数 $A, B$ を用いて

$$
\frac{1}{(p+qx)(1-x)} = \frac{A}{p+qx} + \frac{B}{1-x}
$$

とおく。両辺に $(p+qx)(1-x)$ を掛けると

$$
1 = A(1-x) + B(p+qx)
$$

を得る。右辺を $x$ について整理すると

$$
1 = (A + Bp) + (-A + Bq)\,x
$$

となる。これが任意の $x$ で成り立つためには、定数項と $x$ の係数がそれぞれ両辺で一致しなければならない。すなわち

$$
A + Bp = 1, \qquad -A + Bq = 0
$$

という連立方程式が得られる。第2式より $A = Bq$ であるから、これを第1式に代入すると

$$
Bq + Bp = 1 \quad\Longrightarrow\quad B(p+q) = 1 \quad\Longrightarrow\quad B = \frac{1}{p+q}
$$

となる。これを $A = Bq$ に戻すと

$$
A = \frac{q}{p+q}
$$

を得る。したがって部分分数分解は

$$
\frac{1}{(p+qx)(1-x)} = \frac{q}{p+q}\cdot\frac{1}{p+qx} + \frac{1}{p+q}\cdot\frac{1}{1-x}
$$

となる。この形にしておくことで、次節の積分がそれぞれ対数関数に帰着し、解析的に実行できるようになる。


## 4. 積分と初期条件

両辺を積分すると

$$

\int \frac{dx}{(p+qx)(1-x)} = \int dt

$$

すなわち

$$

\frac{1}{p+q}\ln(p+qx) - \frac{1}{p+q}\ln(1-x) = t + C

$$

である。よって

$$

\frac{1}{p+q}\ln \frac{p+qx}{1-x} = t + C

$$

両辺に $p+q$ を掛けて指数を取れば

$$

\frac{p+qx}{1-x} = C_1 e^{(p+q)t}

$$

となる。初期条件 $N(0)=0$ は $x(0)=0$ と同値なので、$t=0, x=0$ を代入すると $C_1=p$ である。

したがって

$$

\frac{p+qx}{1-x} = p e^{(p+q)t}

$$

を得る。


## 5. 解

上の式を $x$ について解くと

$$

x(t) = \frac{p\left(e^{(p+q)t}-1\right)}{pe^{(p+q)t}+q}

$$

である。分子分母に $e^{-(p+q)t}$ を掛けると

$$

x(t) = \frac{1-e^{-(p+q)t}}{1+\frac{q}{p}e^{-(p+q)t}}

$$

となる。$N(t)=Mx(t)$ を戻せば

$$

N(t)=M \cdot \frac{1-e^{-(p+q)t}}{1+\frac{q}{p}e^{-(p+q)t}}

$$

となり、式 (3.10) が得られる。
